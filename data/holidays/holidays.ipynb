```python
import pandas as pd
import numpy as np
```

```python
# get all Bundesländer
bundeslaender = pd.read_csv("../population_bundesland.csv", sep=";")["Bundesland"].values
```

# Gesetzliche Feiertage

```python
def get_holiday_csv(year_string_in):
    # read holiday xslx file
    holidays = pd.read_excel('holidays.xlsx', sheet_name=year_string_in)

    # create pandas dataframe with all dates in 2020 as index and all Bundesländer as columns
    dates = pd.date_range(start=f'{year_string_in}-01-01', end=f'{year_string_in}-12-31')
    df = pd.DataFrame(index=dates, columns=bundeslaender)

    # set all values to 0
    df = df.fillna(0)

    # iterate through all rows of holidays_2020
    # to parse the dates in "Datum"
    # and Bundesländer in "bundesländer"
    # and set the corresponding values in df_2020 to 1
    for index, row in holidays.iterrows():
        if row["Datum"] is not np.nan:
            # get date
            date_str = row["Datum"].split(" ")[0]
            date = pd.to_datetime(date_str, format="%d.%m.%Y")

            # get bundesländer
            bundeslaender_str = row["Bundesländer"]
            ## replace " und" with ","
            bundeslaender_str = bundeslaender_str.replace(" und", ",")
            ## split at ","
            bundeslaender_strings = bundeslaender_str.split(", ")

            for string in bundeslaender_strings:
                if string == "Alle Bundesländer":
                    for bundesland in bundeslaender:
                        df.loc[date, bundesland] = 1
                else:
                    df.loc[date, string] = 1

    # save dataframe as csv
    df.to_csv(f'holidays_{year_string_in}.csv')

get_holiday_csv("2020")
get_holiday_csv("2021")
```

```python
# put all holidays in one csv
df_2020 = pd.read_csv("holidays_2020.csv", index_col=0)
df_2021 = pd.read_csv("holidays_2021.csv", index_col=0)
df = pd.concat([df_2020, df_2021])
df.to_csv("holidays.csv")
```

# Schulferien

```python
# create null pandas data frame with all dates in 2020 and 2021 as index and all Bundesländer as columns
dates = pd.date_range(start="2020-01-01", end="2021-12-31")
df = pd.DataFrame(index=dates, columns=bundeslaender)
df = df.fillna(0)
```

Get christmas holidays 2019/2020.

```python
holidays = pd.read_excel('school_holidays.xlsx', sheet_name="2019", index_col=0)

# iterate through all rows of holidays and columns of holidays
# to parse the dates of the holidays
# and set the corresponding values in df to 1
for index, row in holidays.iterrows():
    column = "Weihnachtsferien"
    # get date
    string = str(row[column])
    date_strings = string.split(" - ")
    start_date = pd.to_datetime('01.01.2020', format="%d.%m.%Y")
    end_date = pd.to_datetime(date_strings[1] + ".2020", format="%d.%m.%Y")
    dates = pd.date_range(start=start_date, end=end_date)
    for date in dates:
        df.loc[date, index] = 1
```

Get 2020 holidays.

```python
year_string = "2020"
holidays = pd.read_excel('school_holidays.xlsx', sheet_name=year_string, index_col=0)

# iterate through all rows of holidays and columns of holidays
# to parse the dates of the holidays
# and set the corresponding values in df to 1
for index, row in holidays.iterrows():
    for column in holidays.columns:
        if row[column] is not np.nan:
            # get date
            strings = str(row[column]).split(", ")
            for string in strings:
                if " - " in string:
                    date_strings = string.split(" - ")
                    start_date = pd.to_datetime(date_strings[0] + "." + year_string, format="%d.%m.%Y")
                    end_date = pd.to_datetime(date_strings[1] + "." + year_string, format="%d.%m.%Y")
                    dates = pd.date_range(start=start_date, end=end_date)
                    for date in dates:
                        df.loc[date, index] = 1
                else:
                    date = pd.to_datetime(string + "." + year_string, format="%d.%m.%Y")
                    df.loc[date, index] = 1
```

Get 2021 holidays.

```python
year_string = "2021"
holidays = pd.read_excel('school_holidays.xlsx', sheet_name=year_string, index_col=0)

# iterate through all rows of holidays and columns of holidays
# to parse the dates of the holidays
# and set the corresponding values in df to 1
for index, row in holidays.iterrows():
    for column in holidays.columns:
        if row[column] is not np.nan:
            # get date
            strings = str(row[column]).split(", ")
            for string in strings:
                if " - " in string:
                    date_strings = string.split(" - ")
                    if date_strings[1][-2:] == "01":
                        date_strings[1] = "31.12"
                    start_date = pd.to_datetime(date_strings[0] + "." + year_string, format="%d.%m.%Y")
                    end_date = pd.to_datetime(date_strings[1] + "." + year_string, format="%d.%m.%Y")
                    dates = pd.date_range(start=start_date, end=end_date)
                    for date in dates:
                        df.loc[date, index] = 1
                else:
                    date = pd.to_datetime(string + "." + year_string, format="%d.%m.%Y")
                    df.loc[date, index] = 1
```
