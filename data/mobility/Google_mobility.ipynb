```python
import numpy as np
import pandas as pd
```

```python
mobility_DE = pd.read_csv("google_mobility_DE.csv")

states = mobility_DE["sub_region_1"].unique()[1:]
```

```python
state_data = mobility_DE[mobility_DE["sub_region_1"] == states[0]]
mobility_data_column = "residential_percent_change_from_baseline"
```

```python
# make new df with states as columns and dates as rows
mobility_data = pd.DataFrame(columns=states, index=state_data["date"])

# fill in with mobility data from mobility_data_column
# by iterating over mobility_DE
for state in states:
    state_data = mobility_DE[mobility_DE["sub_region_1"] == state]
    mobility_data[state] = state_data[mobility_data_column].values

# cut off after index date "2021-12-31"
mobility_data = mobility_data.loc[: "2021-12-31"]

# save to csv
mobility_data.to_csv("mobility_data_by_Bundesland.csv")
```
